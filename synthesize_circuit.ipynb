# 1. Load an AlphaCNOT model

```python
import sys
import os
from scripts.utils import circuit_to_matrix, apply_AlphaCNOT
from scripts.utils import load_alphacnot_model
import jax.numpy as jnp

# Choose your cardinality
card = 7
# Choose your topology, default ("None") is fully connected
topology = "ALL"
# Choose your source (trained_models or user)
source = "trained_models"
# Choose the reward type (mix or inf for fully connected pretrained models)
reward_type = "mix"
# Choose identity or permutation-relaxed terminal mode
terminal_mode = "identity"

model = load_alphacnot_model(
    card, topology, source, reward_type, terminal_mode
)
```

# 2. Define your list of CNOTs 

```python
cnot_list = [ (2,5), (3,6), (4,0), (6,5), (0,1), (1,4), (1,2), (2,1), (0,1), (4,3), (5,6), (0,6), (0,5), (5,4), (5,4)]

print("Initial number of CNOT gates: ", len(cnot_list))
```

# 3. Visualization of the associated parity matrix

```python
matrix = circuit_to_matrix(cnot_list, card)
print(f"Selected matrix: \n{matrix}")
```

# 4. Synthesis found by AlphaCNOT

```python
print(f"Initial CNOT list ({len(cnot_list)} CNOT): {cnot_list}")

new_cnot_list, solved = apply_AlphaCNOT(model, cnot_list, card)

print(f"New CNOT list ({len(new_cnot_list)} CNOT): {new_cnot_list}")
print(f"Solved: {solved}")
```

# 5. Last check

```python
# Check if the new CNOT list corresponds to an equivalent matrix
new_matrix = circuit_to_matrix(new_cnot_list, card)
print(f"New matrix: \n{new_matrix}")
print(f"Is the new matrix equivalent to the initial matrix?\n{jnp.allclose(new_matrix, matrix, atol=1e-5)}")
```
